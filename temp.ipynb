```python
import threading

lock = threading.Lock()

def foo():
    print("Acquiring lock first time")
    lock.acquire()
    
    print("Trying to acquire lock second time")
    lock.acquire()  # DEADLOCK: will block forever

    print("Inside critical section")
    lock.release()
    lock.release()
```

```python
def subArrDivKatLeast2(nums, k):
    moudulo_First_Ind = dict()
    moudulo_First_Ind[0] = -1
    prefixSum = 0
    for idx, num in enumerate(nums):
        prefixSum += num
        check = prefixSum % k
        if(check in moudulo_First_Ind and idx - moudulo_First_Ind[check] > 1):
            return True
        if check not in moudulo_First_Ind:
            moudulo_First_Ind[check] = idx
    return False
```

```python
print(subArrDivKatLeast2([3,3], 6))
```
